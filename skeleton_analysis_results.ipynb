# Display standard dataset embryo results

```python
save=False
#Get the analytics data
summary_df=analysis.initialise_summary()

#Use "get_embryo_IDs_from_drug" to provide an array of embryo IDs of the dataset you want to view
#No arguments = normal set
normal_set = database.get_embryo_IDs_from_drug()

#Set up path to save results
save_path = results_path / "skeleton_analysis_results" / "main"

#Plot all of the features
for i in range(len(summary_df.columns)):
    feature=summary_df.columns[i]
    title = titles[i]
    if save:
        analysis.plot_feature_by_stage(feature, title, normal_set,save_path=save_path)
    else:
        analysis.plot_feature_by_stage(feature, title, normal_set)
```
