```python
print("Loading packages")
import cv2

import tifffile as tiff
import pickle

from algorithms.visualisation import image_plot,nodes_plot
from algorithms.analysis import gen_networkx_graph
from algorithms import skeleton_model
from algorithms import database

from config import main_image_path
```

```python
stage = 5
n=1

image = tiff.imread(main_image_path / f'hh{stage}_n{n} skeleton.tif')
dists = tiff.imread(main_image_path / f'hh{stage}_n{n} distmap.tif') 


height = len(image)
width = len(image[0])
print(f"Dimensions in pixels {width}x{height}")

#set up node and edge datastructures
#From first principles so we can visualise the diff between unmerged/merged
nodes,adj = skeleton_model.nodes_edges_from_image(image,dists)
print(f"Unmerged length:{len(nodes)}")


#apply merging on nearby nodes
nodes2,adj2 = skeleton_model.merge_nearby_nodes(nodes,adj)
print(f"Merged length:{len(nodes2)}")
```

Unmerged

```python
original = cv2.imread(main_image_path / f'hh{stage}_n{n} plain_scaled.jpg')
skele = tiff.imread(main_image_path / f'hh{stage}_n{n} skeleton.tif')
dists = tiff.imread(main_image_path / f'hh{stage}_n{n} distmap.tif') 

size = [300,300]
x_min = 1000
y_min = 800

image_plot(original,size,x_min,y_min)
image_plot(skele,size,x_min,y_min)
image_plot(dists,size,x_min,y_min)


#Display the skeleton model over the original image
nodes_plot(original,nodes,adj,size,node_alpha=1,x_min=x_min,y_min=y_min,node_scaling=10)
#Display node weights inherited from the distance map
nodes_plot(dists,nodes,adj,size,edge_alpha=1,node_alpha=0.8,node_weights=True,x_min=x_min,y_min=y_min)
#Display the edges and weights on a black background
nodes_plot(None,nodes,adj,size,node_alpha=1,edge_weights=True,x_min=x_min,y_min=y_min)
```
